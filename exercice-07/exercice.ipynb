```python
import seaborn as sns
import pandas as pd

# Chargement du jeu de données Titanic
titanic = sns.load_dataset("titanic")

# 1. Nombre de passagers mineurs (< 18 ans)
mineurs = titanic[titanic["age"] < 18]
nb_mineurs = len(mineurs)
print(f"Moins de 18 ans : {nb_mineurs}")

# 2. Les 5 billets les plus chers avec la classe et le port d'embarquement
top_5_billets = titanic.sort_values("fare", ascending=False)[["fare", "class", "embark_town"]].head(5)
billet_plus_cher = titanic["fare"].max()
print(f"Billet le plus cher : {billet_plus_cher:.2f}")
print("\nTop 5 des billets les plus chers :")
print(top_5_billets)

# 3. Part des femmes parmi les passagers de 1re classe
passagers_1re = titanic[titanic["pclass"] == 1]
part_femmes_1re = (passagers_1re["sex"] == "female").mean() * 100
print(f"\nFemmes en 1re classe : {part_femmes_1re:.1f} %")

# 4. Taux de survie comparés
# Femmes de 1re classe
femmes_1re = titanic[(titanic["pclass"] == 1) & (titanic["sex"] == "female")]
taux_survie_femmes_1re = femmes_1re["survived"].mean() * 100
print(f"Survie femmes 1re : {taux_survie_femmes_1re:.1f} %")

# Hommes de 3e classe
hommes_3e = titanic[(titanic["pclass"] == 3) & (titanic["sex"] == "male")]
taux_survie_hommes_3e = hommes_3e["survived"].mean() * 100
print(f"Survie hommes 3e : {taux_survie_hommes_3e:.1f} %")
```

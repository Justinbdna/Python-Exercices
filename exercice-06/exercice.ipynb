```python
import seaborn as sns
import pandas as pd

# Chargement du jeu de données Titanic
titanic = sns.load_dataset("titanic")

# 1. Dimensions du dataset
print("--- 1. Dimensions ---")
print(f"Lignes et colonnes : {titanic.shape}")

# 2. Types des colonnes
print("\n--- 2. Types des colonnes ---")
print(titanic.dtypes)

# 3. Valeurs manquantes par colonne
print("\n--- 3. Valeurs manquantes ---")
manquants = titanic.isna().sum()
print(manquants[manquants > 0])

# 4. Répartition des passagers par classe et par port
print("\n--- 4. Répartition par classe ---")
print(titanic["class"].value_counts())

print("\n--- Répartition par port d'embarquement ---")
print(titanic["embark_town"].value_counts())
```

### 5. Trois constats d'inspection

1. **Valeurs manquantes critiques** :
   - La variable `deck` présente un manque massif avec **688 valeurs manquantes sur 891** (soit plus de 77 % d'absence), ce qui la rend inexploitable telle quelle sans stratégie d'imputation ou de suppression.
   - La variable `age` comporte également **177 valeurs manquantes** (~20 % de l'échantillon).

2. **Redondance des colonnes** :
   - Plusieurs paires de variables doublonnent la même information : `survived` (numérique 0/1) et `alive` (texte 'yes'/'no'), `pclass` (1/2/3) et `class` ('First'/'Second'/'Third'), ainsi que `embarked` ('S', 'C', 'Q') et `embark_town` ('Southampton', 'Cherbourg', 'Queenstown').
   - La colonne `alone` est une déduction logique directe (`sibsp == 0` et `parch == 0`).

3. **Structure sociologique et géographique** :
   - Plus de la moitié des passagers voyageaient en **3e classe (491 passagers, soit 55 %)** contre 216 en 1re et 184 en 2e.
   - Le port principal d'embarquement était **Southampton (644 passagers, soit 72 %)**.
